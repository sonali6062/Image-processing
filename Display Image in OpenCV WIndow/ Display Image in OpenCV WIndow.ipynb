# Displaying 3 Frames on RGB

```python
img = cv.imread('Transition of colors.jpg')

img = cv.cvtColor(img, cv.COLOR_BGR2RGB)

r = img[ :  , : , 0 ]  # - Red
g = img[ :  , : , 1 ]  # - Green
b = img[ :  , : , 2 ]  # - Blue


cv.imshow("Frame",r)
cv.waitKey(0)

cv.imshow("Frame",g)
cv.waitKey(0)

cv.imshow("Frame",b)
cv.waitKey(0)
```
