```python
import cv2 as cv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Loading an image using OpenCV

```python
img=cv.imread("Transition of colors.jpg")
plt.imshow(img)
```

# Converting RGB to BGR

```python
img=cv.cvtColor(img,cv.COLOR_RGB2BGR)
plt.imshow(img)
```

# Frame Extraction 

```python
r=img[:,:,0]#red
g=img[:,:,1]#green
b=img[:,:,2]#blue
```

```python
plt.imshow(r)
#displays only the area containing red
```

```python
img=cv.cvtColor(r,cv.COLOR_RGB2BGR)
plt.imshow(img)
```

```python
plt.imshow(g)
#displays only the area containing green
```

```python
img=cv.cvtColor(g,cv.COLOR_RGB2BGR)
plt.imshow(img)
```

```python
plt.imshow(b)
#displays only the area containing blue
```

```python
img=cv.cvtColor(b,cv.COLOR_RGB2BGR)
plt.imshow(img)
```

```python
cv.imwrite("red.jpg",r)
cv.imwrite("green.jpg",g)
cv.imwrite("blue.jpg",b)
```
