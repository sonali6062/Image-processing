# Frame Extraction

RGB to HSV(Hue, saturation, and value color model)

```python
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    hsv=cv.cvtColor(img,cv.COLOR_RGB2HSV)#hue shows what is the main image color that's domination
    cv.imshow("Frame",img)
    cv.imshow("HSV",hsv)
    key=cv.waitKey(20)
    if(key==27):
        cam.release()
        break
    
```

```python
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    r=img[:,:,0]
    g=img[:,:,1]
    b=img[:,:,2]
    hsv=cv.cvtColor(img,cv.COLOR_RGB2HSV)#hue shows what is the main image color that's domination
    h=hsv[:,:,0]
    s=hsv[:,:,1]
    v=hsv[:,:,2]
    cv.imshow("Frame",img)
    cv.imshow("Frame Red",r)
    cv.imshow("Frame Green",g)
    cv.imshow("Frame Blue",b)
    cv.imshow("HSV",hsv)
    cv.imshow("H",h)
    cv.imshow("S",s)
    cv.imshow("V",v)
    key=cv.waitKey(20)
    if(key==27):
        cam.release()
        break
  
```
