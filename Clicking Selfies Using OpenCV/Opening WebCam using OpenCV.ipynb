# Working with WebCam

```python
# cam=cv.VideoCapture(1)#0 denotes the default webcam --if external camera than 1
#video is made up of frames
#cam.release()-- to release camera
# Open camera (0 = default webcam, 1 = external)
import cv2 as cv
while True:
    cam = cv.VideoCapture(0)  # 0 = default webcam, 1 = external #smaller the value better the pixel capture
    _,img=cam.read()
    cv.imshow("Frame",img)
    
    key=cv.waitKey(1)#1000-1 sec
    if(key==27): #work on ascii --esc key
        cam.release()
        break

```

```python

cam.release()
cv.destroyAllWindows()
```
