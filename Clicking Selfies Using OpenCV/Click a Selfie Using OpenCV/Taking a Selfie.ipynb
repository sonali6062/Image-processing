```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Clicking Selfie

```python
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    cv.imshow("Frames",img)
    key=cv.waitKey(20)
    if(key==13):#enter 
        cv.imwrite("Selfie.png",img)
        cam.release()
        break
```

# Clicking Selfie(RGB,GrayScale,HSV) at the real time

```python
import cv2 as cv
cam = cv.VideoCapture(0)

while True:

    _, img = cam.read()
    
    img = cv.flip(img, 1) 
    
    gray = cv.cvtColor(img, cv.COLOR_RGB2GRAY)
    hsv  = cv.cvtColor(img, cv.COLOR_RGB2HSV)
   
    cv.imshow("Frame",img)
    
    key = cv.waitKey(20)
    
    if (key == 13):
        
        cv.imwrite('Selfie.png', img)
        cv.imwrite('Selfie_Gray.png', gray)
        cv.imwrite('Selfie_HSV.png', hsv)
        cam.release()
        
        break
```
