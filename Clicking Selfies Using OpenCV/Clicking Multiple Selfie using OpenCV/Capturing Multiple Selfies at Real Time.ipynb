```python
import cv2 as cv
import pandas as pg
import matplotlib.pyplot as plt
```

```python
cam=cv.VideoCapture(0)
c=1
while True:
    _,img=cam.read()
    # img=cv.flip(img,1)
    cv.imshow("Frame",img)
    key=cv.waitKey(30)
    if(key==13): # Click selfie if Enter is pressed
        cv.imwrite("Selfie_"+str(c)+".png",img)
        c+=1
    if(key==27): # Terminate the program if Escape is Pressed
        cam.release()
        break
```
