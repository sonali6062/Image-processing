# Blur the image

=>Averaging Blur=
cv.blur(src, ksize)

ksize = kernel size, e.g. (5,5)


=>Gaussian Blur=
cv.GaussianBlur(src, ksize, sigmaX)

sigmaX = standard deviation in X direction (0 lets OpenCV auto-calc)



=>Median Blur=
cv.medianBlur(src, ksize)

ksize must be an odd integer (3, 5, 7, …)


=>Bilateral Filter
cv.bilateralFilter(src, d, sigmaColor, sigmaSpace)

d = diameter of pixel neighborhood

sigmaColor = how much color difference is allowed

sigmaSpace = how far in space pixels influence each other

```python
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    blur1=cv.blur(img,(2,2))#Kernel(same as cnn) more the kernel size more blur the image
    blur2=cv.blur(img,(4,4))
    blur3=cv.blur(img,(6,6))
    blur4=cv.blur(img,(8,8))
    cv.imshow("Original Frame",img)
    cv.imshow("Blur1 Frame",blur1)
    cv.imshow("Blur2 Frame",blur2)
    cv.imshow("Blur3 Frame",blur3)
    cv.imshow("Blur4 Frame",blur4)
    if cv.waitKey(10)==27:
        cam.release()
        break
cam.release()
    
```

# Blur the image along with Edge detection

```python
import cv2 as cv

cam = cv.VideoCapture(0)

while True:
    ret, img = cam.read(20)
    if not ret:
        break

    img = cv.flip(img, 1)

    # Different blur levels
    # blur1 = cv.blur(img, (6, 6))
    blur2 = cv.blur(img, (8, 8))

    # Edge detection (Canny)
    edges = cv.Canny(img, 100, 200)
    # edges1 = cv.Canny(blur1, 100, 200)
    edges2 = cv.Canny(blur2, 100, 200)

    # Show original + blurs
    cv.imshow("Original Frame", img)
    # cv.imshow("Blur1 Frame", blur1)
    cv.imshow("Blur2 Frame", blur2)
    

    # Show edge detection results
    cv.imshow("Edges original", edges)
    # cv.imshow("Edges Blur1", edges1)
    cv.imshow("Edges Blur2", edges2)
    

    # ESC to exit
    if cv.waitKey(10) == 27:
        break

# cam.release()
# cv.destroyAllWindows()
```

# MedianBlur

# Bilateral Blur

```python
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    bilateral = cv.bilateralFilter(img, 9, 75, 75)



    cv.imshow("Original Frame",img)
    cv.imshow("bilateral Frame",bilateral)
    
    if cv.waitKey(10)==27:
        cam.release()
        break
cam.release()
```

# All blur for better comaparison

```python
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    Canny=cv.blur(img,(8,8))#Kernel(same as cnn) more the kernel size more blur the image
    medianBlur=cv.medianBlur(img, 5)
    gaussian = cv.GaussianBlur(img, (5,5), 0)
    bilateral = cv.bilateralFilter(img, 9, 75, 75)
    cv.imshow("Original Frame",img)
    cv.imshow("Canny Frame",Canny)
    cv.imshow("medianBlur Frame",medianBlur)
    cv.imshow("gaussian Frame",gaussian)
    cv.imshow("bilateral Frame",bilateral)
    if cv.waitKey(10)==27:
        cam.release()
        break
cam.release()
```
