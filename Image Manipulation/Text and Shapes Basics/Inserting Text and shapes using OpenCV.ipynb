# Drawing Rectangle

```python
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    img=cv.rectangle(img,(200,250),(400,450),(150,150,0),10)#img,points,length,width,bgr format,width
    img=cv.imshow("Frame",img)
    key=cv.waitKey(30)
    if(key==27):
        cam.release()
        break
```

# Drawing Circle

```python
import numpy as np
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    img=cv.circle(img,(200,350),50,(150,150,0),10)#img,(center,height), points,radius,bgr format,width
    img=cv.imshow("Frame",img)
    
    key=cv.waitKey(30)
    if(key==27):
        cam.release()
        break

```

# Writing text

```python
import numpy as np
import cv2 as cv
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    img=cv.putText(img,'Sonali',(200,100),cv.FONT_HERSHEY_PLAIN,4,(155,36,89),2)
    cv.imshow("Text Display",img)
    key=cv.waitKey(10)
    if(key==27):
        cam.release()
        break
    
```
