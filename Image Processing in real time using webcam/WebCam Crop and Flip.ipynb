# Flipping an image

```python
import cv2 as cv

cam = cv.VideoCapture(0)  # 0 for default webcam

while True:
    ret, img = cam.read()
    if not ret:
        break  # in case frame not captured

    flip = cv.flip(img,1)  # Flip vertically for 0(complete flip) 1-(mirror effect)

    cv.imshow("Frame", img)
    cv.imshow("Flipped", flip)

    key = cv.waitKey(1)
    if key == 27:  # ESC key to exit
        break

cam.release()
cv.destroyAllWindows()
```
