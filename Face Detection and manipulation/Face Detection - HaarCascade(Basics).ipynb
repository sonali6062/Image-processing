```python
import cv2 as cv
import numpy as np
```

```python
classifier=cv.CascadeClassifier("haarcascade_frontalface_default.xml")
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    faces=classifier.detectMultiScale(img,1.1,8)
    cv.imshow("Frame",img)
    print(faces)
    print('-'*20)
    if cv.waitKey(1)==27:
        cam.release()
        break
```

 the 4 values represent- coordinates of the face(top-left x axis,top-left y axis,width,height of frame)

```python
classifier=cv.CascadeClassifier("haarcascade_frontalface_default.xml")
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    faces=classifier.detectMultiScale(img,2,2)#img,confidence,neighbours(higher the confidence less faces wil be detected
    #lower the value, lower computational cost(lower the value more faces will be detected-can give wrong result)
    for (x,y,w,h) in faces:
        cv.rectangle (img,(x,y),(x+w,y+h),(250,180,0),5)
    cv.imshow("Frame",img)
    print(faces)
    print('-'*20)
    if cv.waitKey(1)==27:
        cam.release()
        break
```

For crowded places like malls, we'll have to detect multiple faces that will be very small, and for better accuracy, we'll have to detect them all.So there we'll work on different parameters
