```python
import cv2 as cv
import numpy as np

# Read image
img=cv.imread("group.jpg")

# Convert to grayscale
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# Load classifier
classifier = cv.CascadeClassifier('haarcascade_frontalface_default.xml')

# Detect faces
faces = classifier.detectMultiScale(gray, 1.1, 5)

# Draw rectangles & show cropped faces
for (x, y, w, h) in faces:
    # Draw rectangle around face
    cv.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)

    # Crop the face
    face = img[y:y+h, x:x+w]

    # Show cropped face
    cv.imshow('Face', face)
    cv.waitKey(0)

# Show final image with rectangles
cv.imshow('Frame', img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import os
import cv2 as cv
import numpy as np

# Load image
img=cv.imread("group.jpg")
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# Load Haar Cascade
classifier = cv.CascadeClassifier('haarcascade_frontalface_default.xml')
faces = classifier.detectMultiScale(gray, 1.1, 5)

# Ensure output folder exists
output_folder = "People"
os.makedirs(output_folder, exist_ok=True)

def save(frame, folder_name):
    name_img = len(os.listdir(folder_name)) + 1
    name_img = os.path.join(folder_name, f"IMG_{name_img}.jpeg")
    cv.imwrite(name_img, frame)
    print(name_img, "is exported")

# Loop over detected faces
for (x, y, w, h) in faces:
    face = img[y:y+h, x:x+w]
    cv.imshow('Face', face)

    # Wait for key press (ENTER=Save, DELETE=Skip, ESC=Exit)
    key = cv.waitKey(0)

    if key == 13:   # ENTER key
        save(face, output_folder)
    elif key == 127:  # DELETE key
        print("Skipped")
    elif key == 27:   # ESC key → exit loop
        break

cv.destroyAllWindows()
 
```
