```python
# Load the Haar Cascade Classifier for face detection
classifier = cv.CascadeClassifier('haarcascade_frontalface_default.xml')

# Start video capture from the default camera (webcam)
cam = cv.VideoCapture(0)

while True:
    
    # Read one frame from the camera
    _, img = cam.read()
    
    # Flip the frame horizontally (mirror effect)
    img = cv.flip(img, 1)
    
    # Detect faces in the frame (returns list of bounding boxes)
    # Parameters: scaleFactor=1.1, minNeighbors=5
    faces = classifier.detectMultiScale(img, 1.1, 5)
    
    # Loop through detected faces
    for f in faces:                             
        if f[-1] == max(faces[:,-1]):           # Choose the face with the largest area (last value = width/height)
            break 

    # If at least one face is detected
    if (len(faces) >= 1):                       
        x = f[0]    # x-coordinate of top-left corner
        y = f[1]    # y-coordinate of top-left corner
        w = f[2]    # width of bounding box
        h = f[3]    # height of bounding box

        # Draw a green rectangle around the detected face
        cv.rectangle(img, (x,y),(x+w,y+h) , (0,180,0), 2)   
        
        # Crop the face region from the frame
        face = img[y:y+h, x:x+w]
        
        # Resize the cropped face to 256x256 pixels
        face = cv.resize(face, (256,256))

    # Show the full video frame
    cv.imshow('Frame', img)
    
    # Show only the cropped face
    cv.imshow('Face', face)
    
    # Exit when 'Esc' key (27) is pressed
    if cv.waitKey(1) == 27:
        cam.release()  # Release the webcam
        break
```
