```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
import cv2 as cv
lower=np.array([94,64,41])
upper=np.array([200,220,240])
img=cv.imread('img1.jpg')
mask = cv.blur(img, (2,2))
mask = cv.inRange(mask, lower, upper)
cv.imshow("Frame-1" , img)
cv.imshow("Mask-1"  , mask)

cv.waitKey(0)
```
