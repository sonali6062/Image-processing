# Masking a specific color in Live video feed

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
```

```python
import cv2 as cv
import numpy as np
camera = cv.VideoCapture(0)

lower = np.array([100,113,113])
upper = np.array([160,160,160])

while True:
    
    _ , img = camera.read()
    img = cv.flip(img,1)    
    
    mask = cv.blur(img, (4,4))
    mask = cv.inRange(mask, lower, upper)
    
    cv.imshow("Frame" , img)
    cv.imshow("Mask"  , mask)
    
    if (cv.waitKey(1) == 27):
        camera.release()
        break
        
camera.release()          
```

When you work with masking in OpenCV, the mask itself is usually expected to be a single-channel (grayscale) image, not in BGR format.for making find out the color value using digital color meter and fed it. For more clear image blur the image first and then apply masking
