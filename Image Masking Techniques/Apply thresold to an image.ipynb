```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

cam = cv.VideoCapture(0)
while True:
    _, img = cam.read()
    
    img = cv.resize(img, (int(1080/2), int(700/2)))
    img = cv.flip(img, 1)
    gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
    
    # FIX: use NumPy indexing instead of function call
    gray_thereshold_1 = gray.copy()
    gray_thereshold_1[gray_thereshold_1 >127] = 200
    gray_thereshold_2 = gray.copy()
    gray_thereshold_2[gray_thereshold_2 <= 127] = 0  # set other pixels to black
    
    cv.imshow("Oiginal Image", img)
    cv.imshow("GrayScale Image", gray)
    cv.imshow("gray_thereshold_1 Image", gray_thereshold_1)
    cv.imshow("gray_thereshold_2 Image", gray_thereshold_2)
    
    if cv.waitKey(1) == 27:  # ESC key
        cam.release()
        cv.destroyAllWindows()   # FIX: close windows
        break
```
