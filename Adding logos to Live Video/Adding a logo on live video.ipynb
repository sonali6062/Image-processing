```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    cv.imshow("Frame",img)
    if cv.waitKey(1)==27:
        cam.release()
        break
```

```python
logo=cv.imread("a.jpg")
plt.imshow(logo)
plt.axis("off")  # optional, removes axes
plt.show()
```

```python
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    img[50:250,400:600]=logo
    cv.imshow("Frame",img)
    if cv.waitKey(1)==27:
        cam.release()
        break
```

```python
logo=cv.resize(logo,(100,100))
logo.shape
```

```python
cam=cv.VideoCapture(0)
while True:
    _,img=cam.read()
    img=cv.flip(img,1)
    img[50:150,500:600]=logo
    cv.imshow("Frame",img)
    if cv.waitKey(1)==27:
        cam.release()
        break
```
