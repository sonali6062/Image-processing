# Top Right

```python
import cv2 as cv
import numpy as np

# Open webcam
cam = cv.VideoCapture(0)

# Load logo
logo = cv.imread('a.jpg')

while True:
    ret, img = cam.read()
    if not ret:
        break

    # Flip horizontally (mirror effect)
    img = cv.flip(img, 1)

    # Get frame dimensions
    ht, wt = img.shape[:2]

    # Logo size & position
    margin_top  = int(ht / 100)       # 1% of frame height
    logo_height = int((ht / 100) * 4) # 4% of frame height
    margin_left = int((wt / 100) * 90) # 90% of frame width
    logo_width  = int((wt / 100) * 4)  # 4% of frame width

    # Resize logo to fit
    logo_resized = cv.resize(logo, (logo_width, logo_height))

    # ---------------------------
    # OPTION 1: Direct Placement
    # ---------------------------
    # img[ margin_top : margin_top + logo_height,
    #      margin_left : margin_left + logo_width ] = logo_resized

    # ---------------------------
    # OPTION 2: Transparent Watermark
    # ---------------------------
    roi = img[margin_top : margin_top + logo_height,
              margin_left : margin_left + logo_width]

    alpha = 0.5  # transparency factor (0 = invisible, 1 = solid)
    blended = cv.addWeighted(roi, 1 - alpha, logo_resized, alpha, 0)

    img[margin_top : margin_top + logo_height,
        margin_left : margin_left + logo_width] = blended

    # Show output
    cv.imshow('Frame', img)

    # Exit on ESC key
    if cv.waitKey(1) == 27:
        break

cam.release()
cv.destroyAllWindows()
```

# Logo at each cornor of frame

```python
import cv2 as cv
import numpy as np
 
cam = cv.VideoCapture(0)

logo = cv.imread('a.jpg')

while True:
    
    _, img = cam.read()

    img = cv.flip(img,1)
    
    ht = img.shape[0]
    wt = img.shape[1]
    
#    Top Right
    margin_top  = int(ht/100)             # 1%
    logo_height = int((ht/100)*7)         # 7%
    margin_left = int((wt/100)*95.5)      # 95.5%
    logo_width  = int((wt/100)*4)         # 4%
    logo = cv.resize(logo, (logo_width, logo_width))
    img[ margin_top : logo_width + margin_top , margin_left : margin_left + logo_width ] = logo
    
#   Top Left
    margin_top  = int(ht/100)             # 1%
    logo_height = int((ht/100)*7)         # 7%
    margin_left = int((wt/100)*.5)        # .5%
    logo_width  = int((wt/100)*4)         # 4%
    logo = cv.resize(logo, (logo_width, logo_width))
    img[ margin_top : logo_width + margin_top , margin_left : margin_left + logo_width ] = logo
    
#   Bottom Left
    margin_top  = int((ht/100)*92)        # 92%
    logo_height = int((ht/100)*7)         # 7%
    margin_left = int((wt/100)*.5)        # .5%
    logo_width  = int((wt/100)*4)         # 4%
    logo = cv.resize(logo, (logo_width, logo_width))
    img[ margin_top : logo_width + margin_top , margin_left : margin_left + logo_width ] = logo

#   Bottom Right
    margin_top  = int((ht/100)*92)        # 92%
    logo_height = int((ht/100)*7)         # 7%
    margin_left = int((wt/100)*95)        # .5%
    logo_width  = int((wt/100)*4)         # 4%
    logo = cv.resize(logo, (logo_width, logo_width))
    img[ margin_top : logo_width + margin_top , margin_left : margin_left + logo_width ] = logo
    
    
    cv.imshow('Frame'  , img)
     
    if cv.waitKey(1) == 27:
        cam.release()
        break
```
