```python
import cv2 as cv

cam = cv.VideoCapture(0)

# Load logo once
logo = cv.imread('a.jpg')
if logo is None:
    print("Error: could not load logo image!")
    cam.release()
    exit()

r = 1280  # initial width crop

while True:
    ret, img = cam.read()
    if not ret:
        break

    img = cv.flip(img, 1)

    # Ensure crop doesn't go out of range
    r = max(100, min(r, img.shape[1]))
    img = img[:, :r]

    ht, wt = img.shape[:2]

    # Logo placement
    margin_top  = int(ht / 100)       # 1% from top
    logo_height = int((ht / 100) * 7) # 7% of height
    margin_left = int((wt / 100) * 95.5) # near right edge
    logo_width  = int((wt / 100) * 4) # 4% of width

    side = min(logo_width, logo_height)
    logo_resized = cv.resize(logo, (side, side))

    # Overlay logo
    img[margin_top : margin_top + side,
        margin_left : margin_left + side] = logo_resized

    cv.imshow('Frame', img)

    # ---- Handle keys (single call) ----
    key = cv.waitKey(1) & 0xFF

    if key == 115:   # 's'
        r -= 10
    elif key == 119: # 'w'
        r += 10
    elif key == 27:  # ESC
        break

cam.release()
cv.destroyAllWindows()
```
